Data Preprocessing


Purpose: Fill missing numeric values with the mean.

```python
df['Price per Night (BDT)'].fillna(df['Price per Night (BDT)'].mean(), inplace=True)
df['Cleanliness Rating'].fillna(df['Cleanliness Rating'].mean(), inplace=True)
df['Service Rating'].fillna(df['Service Rating'].mean(), inplace=True)
df['Food Rating'].fillna(df['Food Rating'].mean(), inplace=True)
df['Safety Rating'].fillna(df['Safety Rating'].mean(), inplace=True)
```

Encode Categorical Variables (like “Location” or “Room Type”)

Purpose: Convert text (Location, Room Type) to numeric codes for machine learning models.

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Location_encoded'] = le.fit_transform(df['Location'])
df['RoomType_encoded'] = le.fit_transform(df['Room Type'])
```

Feature Selection

Example: Price, Cleanliness, Service, Food, Safety, Location, Room Type.

Target → the column the model is trying to predict (output).
Example: User Rating.

Inputs = features that affect User Rating.

Output = what we want to predict (User Rating).

```python
features = ['Price per Night (BDT)', 'Cleanliness Rating', 'Service Rating', 
            'Food Rating', 'Safety Rating', 'Location', 'Room Type']
target = 'User Rating'
```

Purpose: Suggest hotels based on user preferences:

Location

Budget range

Amenities

Minimum user rating

```python
def recommend_hotels(location=None, budget=None, amenities=None, min_rating=0, top_n=3):
    filtered_df = df.copy()
    
    # 1️⃣ Filter by location
    if location:
        filtered_df = filtered_df[filtered_df['Location'].str.contains(location, case=False, na=False)]
    
    # 2️⃣ Filter by budget (min, max)
    if budget:
        low, high = budget
        filtered_df = filtered_df[(filtered_df['Price per Night (BDT)'] >= low) &
                                  (filtered_df['Price per Night (BDT)'] <= high)]
    
    # 3️⃣ Filter by amenities
    if amenities:
        filtered_df['Amenities'] = filtered_df['Amenities'].fillna('')
        for amenity in amenities:
            filtered_df = filtered_df[filtered_df['Amenities'].str.contains(amenity, case=False)]
    
    # 4️⃣ Filter by minimum rating
    filtered_df = filtered_df[filtered_df['User Rating'] >= min_rating]
    
    # 5️⃣ Sort by rating (descending) and price (ascending)
    recommended = filtered_df.sort_values(by=['User Rating','Price per Night (BDT)'], ascending=[False, True])
    
    # 6️⃣ Return top N hotels with selected columns
    return recommended.head(top_n)[['Hotel Name','Location','Price per Night (BDT)','User Rating','Amenities']]
```
